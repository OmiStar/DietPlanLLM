```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
```

```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()


llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
```

```python
goal = input("What is your goal? Gain weight or lose weight? ").strip().lower()

if goal not in ["gain weight", "lose weight"]:
    print("Please enter either 'gain weight' or 'lose weight'.")

else:
    prompt = f"""
    You are a nutrition assistant.

    The user's fitness goal is to {goal}.

    Create a simple and practical daily diet plan for this goal.

    Include:
    - Breakfast
    - Morning snack
    - Lunch
    - Evening snack
    - Dinner
    - Approximate calories for each meal
    - Approximate protein for each meal
    - Total daily calories
    - Total daily protein

    Use common, affordable foods.
    Keep the explanation simple and easy to follow.
    Do not recommend extreme dieting.
    """

    response = llm.invoke(prompt)

    print("\nYour Diet Plan:\n")
    print(response.text)
```
